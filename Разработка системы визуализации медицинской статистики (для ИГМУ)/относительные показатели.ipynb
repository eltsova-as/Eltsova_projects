## 1.	Отчетная форма 8 СВЕДЕНИЯ О ЗАБОЛЕВАНИЯХ АКТИВНЫМ ТУБЕРКУЛЕЗОМ (1000)

```python
# Загружаем библотеку
import pandas as pd
```

```python
# Загружаем файлы

df_1000 = pd.read_csv(r'C:\...\1000_2016-2024.csv')
df_pop = pd.read_csv(r'C:\...\population.csv')
```

```python
# Задаем нужные годы
years_to_filter = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
```

```python
# id 1.1
# Фильтруем числитель
numerator_data = df_1000[
    (df_1000['Таблица'] == 1000) & 
    (df_1000['Строка'].isin([1, 2])) & 
    (df_1000['Графа'] == 5) &
    (df_1000['Год'].isin(years_to_filter))
]

# Группируем по году
numerator = numerator_data.groupby('Год')['Значение'].sum().reset_index()
numerator.columns = ['year', 'numerator']

# Фильтруем знаменатель только для нужных годов
denominator = df_pop[df_pop['year'].isin(years_to_filter)][['year', 'all']].copy()
denominator.columns = ['year', 'denominator']

# Объединяем и рассчитываем
result_1_1 = numerator.merge(denominator, on='year', how='left')
result_1_1['indicator'] = (result_1_1['numerator'] / result_1_1['denominator']) * 100000
result_1_1['indicator'] = result_1_1['indicator'].round(2)

# Добавляем служебные колонки
result_1_1['id'] = 'id_1.1'
result_1_1['name'] = result_1_1['year'].apply(
    lambda x: f'Показатель первичной заболеваемости туберкулезом всего {x} год (на 100 тыс. населения)'
)
result_1_1['short_name'] = 'ПЗ туб тер'

# Итоговый результат
final_result = result_1_1[['id', 'name', 'short_name', 'year', 'indicator', 'numerator', 'denominator']]
print("\nИтоговая таблица :")
print(final_result)
```
